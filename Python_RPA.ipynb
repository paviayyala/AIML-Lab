```python
# Install dependencies
!pip install -q playwright pandas
!playwright install chromium
# Install required system libraries for Playwright Chromium
!apt-get install -y libatk-bridge2.0-0 libatk1.0-0 libgtk-3-0 libgdk-pixbuf2.0-0 libsoup2.4-1 libxcomposite1 libxrandr2 libxtst6 libnss3 libasound2

import asyncio
from playwright.async_api import async_playwright # Changed to async_playwright
import pandas as pd

async def run(): # Made run() an async function
    async with async_playwright() as p: # Changed to async with async_playwright
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        # Step 1: Open login page
        await page.goto("https://the-internet.herokuapp.com/login")

        # Step 2: Login (demo credentials)
        await page.fill("#username", "tomsmith")
        await page.fill("#password", "SuperSecretPassword!")
        await page.click("button[type='submit']")

        # Step 3: Wait for login success
        await page.wait_for_selector(".flash.success")
        print("✅ Logged in successfully")

        # Step 4: Go to table page
        await page.goto("https://the-internet.herokuapp.com/tables")

        # Step 5: Extract table data
        rows = await page.query_selector_all("#table1 tbody tr")

        data = []
        for row in rows:
            cols = await row.query_selector_all("td")
            data.append([await col.inner_text() for col in cols])

        # Step 6: Save to CSV
        df = pd.DataFrame(data, columns=["Last Name", "First Name", "Email", "Due", "Web", "Action"]) # Added "Action" column
        df.to_csv("output.csv", index=False)

        print("✅ Data saved to output.csv")

        await browser.close()

# Run script
await run() # Changed to await run()

# Download file
from google.colab import files
files.download("output.csv")
```
